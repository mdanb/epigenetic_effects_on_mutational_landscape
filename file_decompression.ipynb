```python
# for file decompression
import gzip
import shutil
```

```python
compressed_file = "./data/rawFiles/GSE207938_ProgressionCohort.h5ad.gz"
decompressed_file = "./data/rawFiles/GSE207938_ProgressionCohort.h5ad"

print("File decompression started...")

with gzip.open(compressed_file, 'rb') as f_in:
    with open(decompressed_file, 'wb') as f_out:
        shutil.copyfileobj(f_in, f_out)

print("File decompressed successfully.")
```
